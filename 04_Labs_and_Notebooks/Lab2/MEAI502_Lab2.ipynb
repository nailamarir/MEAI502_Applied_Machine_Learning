# Part 1 — Where Does ML Data Come From?
*≈ 15 minutes · no code*

Before we touch pandas, one mental model. It is the most important idea in this lab.

**When someone hands you data, three separate questions have answers — and beginners collapse
them into one.**

## The three questions

### 1. What is the **source**?

*Where did this information originate? Who or what produced it?*

Government · a hospital · a university · a company · IoT sensors · a mobile application ·
a survey · a social media platform · a laboratory experiment · a public repository

### 2. What was the **acquisition method**?

*How did I get hold of it?*

Downloaded a file · queried a database · called an API · scraped a web page · ran a survey ·
read a sensor · received an internal export from a colleague

### 3. What is the **format**?

*How was it delivered to me?*

CSV · Excel · JSON · a SQL table · Parquet · image files · text documents · audio · video

> **"CSV" is a format, not a source.** Saying *"my data came from a CSV"* answers none of the
> questions that matter. It does not say who produced it, whether you are allowed to use it,
> how current it is, or who is missing from it.

## Two worked examples

Read these across, and notice that the three rows are genuinely independent.

| | Example A | Example B |
|---|---|---|
| **Source** | Saudi government statistics | A national weather service |
| **Acquisition** | Downloaded from an open-data portal | Called a REST API |
| **Format** | CSV | JSON |
| **How current?** | A snapshot — as fresh as the last publication | Live — whatever the server holds now |
| **Reproducible?** | Yes, if you keep the file and its version | Only if you record the query and the date |

**The same source can reach you three different ways.** A weather service might publish a yearly
CSV *and* run an API *and* sell a database connection. Identical measurements, three acquisition
methods, three formats — and three different sets of practical problems.

When you write up a project, state all three:

> *"Daily weather observations for Jeddah (**source**: Open-Meteo, republishing ECMWF ERA5
> reanalysis), **acquired** through their public archive API on 1 September 2026, **delivered**
> as JSON and saved locally as CSV. Licence: CC-BY 4.0."*

That sentence lets someone else get the same data again. That is what reproducibility means.

## The six acquisition methods

| # | Method | What it looks like | Watch out for | Today? |
|---|---|---|---|---|
| 1 | **Download a dataset** | Click a link on a portal or repository | Version and retrieval date — files get updated silently | **Yes** |
| 2 | **Receive it from an organisation** | A colleague emails you an export | No documentation; nobody remembers what the columns mean | **Yes** |
| 3 | **Query a database** | SQL against the organisation's store | You need access; the data spans several tables | **Yes** |
| 4 | **Call an API** | An HTTP request to a live service | Rate limits, quotas, schema changes | **Yes** |
| 5 | **Collect from the web** | Scraping pages built for humans | **Legality and terms of service** | **Yes** |
| 6 | **Sensors, forms, experiments** | You generate the data yourself | Calibration, gaps, consent | Concept only |

### Method 6 is more common than students expect

If you collect data yourself — a survey, a sensor rig, a lab experiment — you own problems nobody
else can fix: who answered and who did not, whether the instrument drifted, whether your
participants consented to *this* use. **You also own the only copy.** Back it up.

## What if the data belongs to a company?

Outside this classroom, the normal workflow is often **not** "go to Kaggle."

It is: **you talk to the person who owns the data.**

Imagine you are building a model for a telecom company that wants to predict which customers will
leave. Nobody hands you a tidy CSV. You get a meeting. **These are the questions to ask in it.**

### The twelve questions

**About the data itself**

1. **What data do you collect?** — you cannot ask for what you do not know exists
2. **Where is it stored?** — a data warehouse, a production database, a vendor's system, spreadsheets on someone's laptop
3. **What time period is available?** — two months of history will not support a seasonal model
4. **What does one row represent?** — one customer? one customer-month? one phone call? **This is the single most important question in the list**
5. **What does each column mean?** — including its units and how it was measured
6. **What is the target?** — and is it actually recorded, or would someone have to label it?

**About how it behaves**

7. **How often is the data generated?** — daily, hourly, streaming, or one annual export?
8. **Are there missing periods?** — a system migration, an outage, a change of vendor. These leave gaps that look like real zeros

**About whether you may use it**

9. **Who is allowed to access it?** — and are you one of them, in writing?
10. **Does it contain personal or sensitive information?** — names, national IDs, phone numbers, health data, location
11. **Can it legally be used for this purpose?** — consent given for billing is not consent for profiling
12. **Can it leave the company's servers?** — this decides whether you may use Colab at all

> **Question 4 catches more projects than any other.** "One row per customer" and "one row per
> customer per month" are different datasets with different targets, different sizes and different
> leakage risks. Ask it early, and ask for an example row.

> **Questions 10–12 are not paperwork.** In Saudi Arabia the **Personal Data Protection Law**
> applies to personal data. Some organisations legally cannot let data leave their network — which
> means no Colab, no laptop copies, and your model must be trained on their infrastructure.
> Find this out in the *first* meeting, not the week before your deadline.

### The answer to question 5 has a name: a **data dictionary**

A data dictionary records what each column means. It is the difference between a file you can use
and a file you are guessing at.

| Column | Meaning | Type | Unit / values |
|---|---|---|---|
| `age` | Customer age at signup | numeric | years |
| `city` | City of residence | categorical | Jeddah, Riyadh, Dammam, … |
| `balance` | Account balance at month end | numeric | SAR |
| `contract_months` | Length of current contract | numeric | months |
| `churn` | Did the customer leave? | Boolean — **target** | 0 = stayed, 1 = left |

**Having the CSV is not enough.** Without this table you do not know whether `balance` is in riyals
or halalas, whether it is the opening or closing balance, or whether `churn = 1` means the customer
left or the customer stayed.

> **Write one for your own project data**, even if nobody gives you one. Building it forces the
> questions above, and it becomes the core of the data card you will write in Lab 3.

# Part 2 — Forms of Data
*≈ 8 minutes · no code*

A second distinction, cutting across the first. It decides how much work stands between you and
a table.

## Structured

**Fixed rows and columns, one defined type per column.**

```
Age | Income | City    | Churn
----+--------+---------+------
31  | 12000  | Jeddah  | 0
45  |  9500  | Riyadh  | 1
```

CSV, Excel, SQL tables, Parquet. **Ready for a DataFrame almost immediately.**

## Semi-structured

**Organised and self-describing, but not a grid.** Nesting is allowed, and records may differ
from one another.

```json
{"customer": {"age": 31, "city": "Jeddah"},
 "plan": {"type": "postpaid", "monthly_sar": 150}}
```

JSON, XML, log files. **Nearly ready** — it must be flattened first.

## Unstructured

**No predefined structure at all.**

Images · audio · video · free text · PDFs · scanned documents

**Not ready.** Numerical features must be extracted before any model can use it.

## The comparison

| Kind | Examples | Ready for a DataFrame? | Where in this course |
|---|---|---|---|
| **Structured** | CSV, Excel, SQL, Parquet | **Yes** | Weeks 2–11 |
| **Semi-structured** | JSON, XML, logs | Nearly — flatten first | Today, and in APIs throughout |
| **Unstructured** | Images, audio, text, video | **No** — extract features first | Week 12 onward |

> **ML data is much broader than tables.** Most students arrive believing a dataset *is* a CSV.
> It is not — most of the world's data is unstructured, and most of the recent progress in AI has
> been about learning useful numerical representations of exactly that.

**How unstructured data eventually becomes usable:**

```
image  ──►  pixels  ──►  numerical array  ──►  tensor  ──►  model
text   ──►  tokens  ──►  numerical vectors ──►  tensor  ──►  model
```

Everything ends up as numbers. The question is only *how* you get there — and for unstructured
data, that conversion **is** most of the problem.

> **A useful rule of thumb.** The more structured your data, the more of your effort goes into the
> model. The less structured it is, the more goes into the representation.

# Part 3 — How Python Represents Data
*≈ 10 minutes*

Machine learning is built on a small hierarchy of mathematical objects. Everything you will meet
this semester is one of these four.

## The hierarchy

```
        One value
            ↓
         SCALAR                    35.2

     Several values
            ↓
         VECTOR                    [35.2, 33.1, 36.8]

     Rows × Columns
            ↓
     MATRIX / TABLE                [[35.2, 62],
                                    [33.1, 58],
                                    [36.8, 65]]

      More dimensions
            ↓
         TENSOR                    1,000 images × 224 × 224 × 3
```

Each level adds one **dimension**. That is the only difference between them.

## The same hierarchy, in Python

| Concept | Example | Python representation | Dimensions |
|---|---|---|---|
| **Scalar** | `35.2` — one temperature | `float`, `int` | 0 |
| **Vector** | Temperature for 7 days | NumPy 1-D array · pandas **Series** | 1 |
| **Matrix** | 100 students × 5 numeric features | NumPy 2-D array | 2 |
| **Table** | Students with names, grades **and** dates | pandas **DataFrame** | 2, mixed types |
| **Tensor** | 1,000 RGB images | NumPy / PyTorch / TensorFlow n-D array | 3+ |

> **Notice the two entries at dimension 2.** A NumPy matrix and a pandas DataFrame are both
> rows × columns — but a matrix holds **one type** of number, while a DataFrame can hold numbers,
> text and dates side by side, with **names** for its columns. That difference is exactly why both
> exist, and Part 4 is about when to use which.

Let us make each one concretely, starting with plain Python.

````python
# A scalar - one value.
temperature = 35.2

print(temperature)
print("type:", type(temperature).__name__)
````

````python
# A vector, as a plain Python list.
temperatures = [35.2, 33.1, 36.8, 34.5, 37.0, 36.2, 35.9]

print(temperatures)
print("type  :", type(temperatures).__name__)
print("length:", len(temperatures))
````

A Python **list** is the everyday way to hold several values. It works, and it is flexible — a list
can hold numbers, text and anything else all at once.

**But it is not built for mathematics.** Watch what happens if we try to double every temperature.

````python
# What does multiplying a LIST by 2 do?
print(temperatures * 2)
````

**It repeated the list instead of doubling the numbers.** Fourteen values, not seven.

That is correct Python — but it is not what anyone doing mathematics wants. To double each value
with a list you would have to write a loop.

**This is precisely the problem NumPy exists to solve**, and it is where Part 4 begins.

# Part 4 — NumPy and pandas
*≈ 12 minutes*

Two libraries do almost all the data work in Python, and they have different jobs.

| Library | Built for | Holds |
|---|---|---|
| **NumPy** | Fast numerical computation | Numbers, all of one type |
| **pandas** | Working with real-world tables | Mixed types, with names and an index |

### Step 4.1 — Import them

These two imports open almost every data science notebook in the world. `np` and `pd` are the
universal abbreviations — use them, so your code reads like everyone else's.

````python
import numpy as np
import pandas as pd

print("numpy ", np.__version__)
print("pandas", pd.__version__)
````

### Step 4.2 — A NumPy array

`np.array()` turns a list into an array.

````python
x = np.array([10, 20, 30, 40])

print(x)
print("type:", type(x).__name__)
````

### Step 4.3 — The three things to ask any array

Every NumPy array can tell you its shape, its number of dimensions, and its type.

````python
print("shape:", x.shape)     # how many elements along each dimension
print("ndim :", x.ndim)      # how many dimensions
print("dtype:", x.dtype)     # what type of number
````

- **`shape` is `(4,)`** — four elements in one dimension. The trailing comma is Python's way of
  writing a one-element tuple; it is not a typo.
- **`ndim` is `1`** — this is a **vector**.
- **`dtype` is `int64`** — 64-bit integers. **Every element has the same type**, which is what
  makes NumPy fast.

### Step 4.4 — Now the arithmetic works

Remember the list that repeated itself? Try the same thing with an array.

````python
print("x      ", x)
print("x * 2  ", x * 2)
print("x + 100", x + 100)
````

**One operation applied to every element, with no loop.** That is the whole idea of NumPy, and it
is why every numerical library in Python is built on top of it.

````python
# Statistics come for free.
print("mean:", x.mean())
print("min :", x.min())
print("max :", x.max())
````

### Step 4.5 — A matrix is the same thing with two dimensions

````python
matrix = np.array([[35.2, 62],
                   [33.1, 58],
                   [36.8, 65]])

print(matrix)
print("\nshape:", matrix.shape)     # (rows, columns)
print("ndim :", matrix.ndim)
````

**`shape` is `(3, 2)` — three rows, two columns.** Always read a shape as *(rows, columns)*.

`ndim` is now `2`, so this is a **matrix**.

> **Shape is the thing you will check most often**, and mismatched shapes are the most common
> error in all of machine learning. Get into the habit now.

### Step 4.6 — But NumPy has a limitation

An array holds **one type**. Real data does not. Watch what happens when we mix text and numbers.

````python
mixed = np.array([["Sara", 23, 88.5],
                  ["Lama", 25, 91.0]])

print(mixed)
print("\ndtype:", mixed.dtype)
````

**Everything became text.** Look at the quote marks: `'23'` and `'88.5'` are now strings, and
`mixed.mean()` would fail.

NumPy forced one type onto all of it, because that is what NumPy is *for*. Real datasets have a
name column, an age column and a date column — all different types.

**That is what pandas is for.**

### Step 4.7 — Build a DataFrame by hand

Before we load any file, let us build a small table ourselves, so you can see exactly what a
DataFrame is.

We start with a **dictionary**: each key becomes a column name, each list becomes that column's
values.

````python
data = {
    "Name":  ["Sara", "Lama", "Noor"],
    "Age":   [23, 25, 22],
    "Score": [88.5, 91.0, 84.5],
}

df = pd.DataFrame(data)

df
````

**A real table.** Three rows, three columns, with column names — and, on the far left, the numbers
`0, 1, 2`, which pandas added automatically. That is the **index**: the row identifier.

Notice that `Name` holds text while `Age` and `Score` hold numbers. **Mixed types, side by side** —
exactly what NumPy could not do.

````python
print("type      :", type(df).__name__)
print("shape     :", df.shape)
print("\ncolumn types:")
print(df.dtypes)
````

### Step 4.8 — A Series is one column

Take a single column out of a DataFrame and you get a **Series**.

````python
ages = df["Age"]

print(ages)
print("\ntype:", type(ages).__name__)
````

**This is the distinction to remember:**

| | What it is | Dimensions |
|---|---|---|
| **Series** | **One** labelled column | 1 |
| **DataFrame** | **Several** labelled columns | 2 |

A DataFrame is a collection of Series that share an index.

````python
# Selecting several columns gives you a DataFrame back.
subset = df[["Age", "Score"]]

print(type(subset).__name__)
subset
````

> **The bracket rule.** `df["Age"]` — single brackets, one name — gives a **Series**.
> `df[["Age", "Score"]]` — double brackets, a *list* of names — gives a **DataFrame**.
>
> Double brackets look strange at first. They are just a list inside the selection.

### Step 4.9 — The vocabulary of a DataFrame

```
                     columns  =  variables  =  FEATURES
                  ┌──────────────────────────────┐
                    Name      Age     Score
        index  0    Sara       23      88.5      ┐
   (row id)   1    Lama       25      91.0      ├─  rows = observations = examples
               2    Noor       22      84.5      ┘
```

| Term | Also called | Meaning |
|---|---|---|
| **Row** | observation, example, instance, sample | One thing you measured — one student, one day, one customer |
| **Column** | variable, feature, attribute | One property measured across all of them |
| **Cell** | value | One measurement |
| **Index** | row label | How pandas identifies each row |

### Step 4.10 — How a DataFrame becomes machine learning input

This is the bridge you will need in every lab from now on.

In Lecture 2 you saw the notation $X$ and $y$. Here is where they come from:

```
                    FEATURES                          TARGET
        ┌────────────────────────────────┐        ┌──────────┐
          age     income     city   ...              churn
  row 1    31      12000    Jeddah                     0
  row 2    45       9500    Riyadh          X          1         y
  row 3    28      15200    Dammam                     0
  ...
        └────────────────────────────────┘        └──────────┘
                  m rows × n features                m rows
```

- **$X$** — the feature matrix. **$m$ rows** (examples) **× $n$ columns** (features)
- **$y$** — the target vector. **$m$ values**, one per row

Which is written:

$$ X \in \mathbb{R}^{m \times n} \qquad y \in \mathbb{R}^{m} $$

> **Read that against the picture.** $X$ is a matrix because it is rows × columns. $y$ is a vector
> because it is one value per row. **$m$ is the number of rows in your DataFrame, and $n$ is the
> number of feature columns** — which is why `df.shape` is the first thing you check.

In Lab 1 you did exactly this without naming it: `X = data.data` and `y = data.target`.

### Step 4.11 — DataFrame vs NumPy array vs tensor

Three ways to hold the same numbers. Each is best at a different moment.

#### DataFrame — best while **understanding and preparing** data

Column names, an index, and mixed types: numbers, text, dates, categories.

This is what you work with today, and in Lab 3.

#### NumPy array — best for **numerical computation**

Homogeneous numbers, no names. This is what you hand to a model.

````python
# A DataFrame becomes a NumPy array with .to_numpy()
X = df[["Age", "Score"]].to_numpy()

print(X)
print("\ntype :", type(X).__name__)
print("shape:", X.shape)
print("dtype:", X.dtype)
````

**The column names are gone.** All that remains is a 3 × 2 grid of numbers — which is precisely
what a learning algorithm wants. `scikit-learn` accepts a DataFrame and does this conversion
internally.

#### Tensor — the deep learning representation

A **tensor** is a multidimensional numerical structure. You will not use one today, but you will
see the word constantly from Week 12, so meet it now:

| What | Shape | Dimensions |
|---|---|---|
| A scalar | `()` | 0 |
| A vector of 5 values | `(5,)` | 1 |
| A matrix, 100 rows × 10 features | `(100, 10)` | 2 |
| One RGB image, 224 × 224 pixels | `(224, 224, 3)` | 3 |
| A batch of 32 such images | `(32, 224, 224, 3)` | 4 |

Read the image shape as *height × width × colour channels* — three channels for red, green and
blue. The batch adds a fourth dimension: *how many images at once*.

> **A tensor is the same idea as a NumPy array, extended to any number of dimensions**, and stored
> in a form that a GPU can process efficiently. PyTorch and TensorFlow are built around them.
>
> **Later, when we train neural networks, our data will ultimately become tensors.**

### Summary of Part 4

| Structure | Use it when | Holds |
|---|---|---|
| Python **list** | Small, quick, general-purpose | Anything — but no maths |
| NumPy **array** | Numerical computation, model input | One type, any dimensions |
| pandas **Series** | One labelled column | One type, with an index |
| pandas **DataFrame** | Understanding and preparing real data | **Mixed** types, named columns |
| **Tensor** | Deep learning | Numbers, many dimensions, GPU-ready |

The usual journey of a dataset:

```
file  ──►  DataFrame  ──►  NumPy array  ──►  (tensor)  ──►  model
           understand      compute                          learn
```

# Part 5 — Data Types
*≈ 8 minutes*

Before loading anything, one more idea: **what kind of thing is each column?**

There are two different answers to that question, and keeping them apart will save you from a
classic beginner's mistake.

## The five semantic types

**What the column *means*.**

### 1. Numerical — a quantity you can do arithmetic on

| Sub-kind | Meaning | Examples |
|---|---|---|
| **Discrete** | Whole things you count | number of children, number of rooms, visits per month |
| **Continuous** | Measured on a scale, any value in a range | temperature, salary, height, time taken |

The test: **does adding two values make sense?** Two temperatures average sensibly. Two customer
ID numbers do not.

### 2. Categorical — a label from a fixed set

| Sub-kind | Meaning | Examples |
|---|---|---|
| **Nominal** | No natural order | city (Jeddah, Riyadh, Dammam), blood type, brand |
| **Ordinal** | **Ordered**, but gaps are not equal | education level, satisfaction (low/medium/high), T-shirt size |

The ordinal distinction matters later: *high* is more than *medium*, but "how much more" is
undefined — so you cannot simply average them.

### 3. Boolean — exactly two values

`True` / `False` · `Yes` / `No` · `0` / `1`

Very common as a **target**: did the customer leave, is the email spam, was the transaction fraud.

### 4. Date and time

`2026-09-22` · `2026-09-22 14:30:00`

Special because they carry hidden structure: a year, a month, a day of the week, a season. Stored
correctly, you can extract all of it. Stored as text, you can extract none of it.

### 5. Text — free-form language

`"The service was excellent but the app kept crashing."`

**Unstructured.** It must be converted to numbers before any model can use it — Week 12.

## How pandas represents them

pandas has its own, shorter list of storage types, and you see them with `.dtypes`:

| pandas dtype | Usually means | Semantic type |
|---|---|---|
| `int64` | Whole numbers | Numerical (discrete) — **or** an ID, or an encoded category |
| `float64` | Decimal numbers | Numerical (continuous) |
| `bool` | `True` / `False` | Boolean |
| `datetime64` | Real dates | Date/time |
| `object` | **Text** — or anything mixed | Categorical or text |
| `category` | Text from a small fixed set | Categorical (memory-efficient) |

## The distinction that matters

> ## The *meaning* of a variable and its pandas *dtype* are related — but they are not the same.

pandas can only see how a value is **stored**. It cannot see what it **means**.

**The classic example:**

A customer ID column stored as `int64`. pandas sees integers, and will happily compute their mean.
But **the average customer ID is a meaningless number.** The column is an *identifier* — a label
that happens to be written with digits.

Other cases of the same mistake:

| Column | Stored as | pandas thinks | Actually is |
|---|---|---|---|
| `customer_id` | `int64` | A quantity | **An identifier** — never average it |
| `postal_code` | `int64` | A quantity | **A category** written in digits |
| `rating_1_to_5` | `int64` | A quantity | **Ordinal** — the gaps may not be equal |
| `is_active` | `int64` (0/1) | A quantity | **Boolean** |
| `date` | `object` | Text | **A date** — no calendar operations until converted |

> **Nobody can fix this for you.** Only a human who understands the problem can say whether a
> column of integers is a quantity, an identifier, or a category. That is what the **data
> dictionary** from Part 1 is for.
>
> **This is an ML lesson, not a pandas lesson.** Feed a customer ID to a model as a number and it
> will dutifully look for a trend in it — and may well find one, purely because IDs were issued in
> chronological order. Your model then "predicts" using the fact that older customers have smaller
> IDs, which is not a discovery. It is a bug.

We will look at real dtypes in Part 7, once we have loaded something.

## 6.6 — APIs

The five sources so far were **files** — snapshots saved at some moment in the past.

An **API** is different. It is a live service that answers a question when you ask it.

```
My Python program
       ↓  request
   API / server
       ↓  response
      JSON
       ↓
   DataFrame
```

### Four words to know

| Term | Meaning | In our example |
|---|---|---|
| **Endpoint** | The address of the service | `archive-api.open-meteo.com/v1/archive` |
| **Parameters** | What exactly you are asking for | latitude, longitude, dates, variables |
| **Request** | The message you send | The full URL below |
| **Response** | What comes back | JSON |

In plain English, our request says:

> *Give me: **the daily maximum temperature**, for **Jeddah**, from **1 June** to **30 June 2025**.*

````python
url = ("https://archive-api.open-meteo.com/v1/archive"
       "?latitude=21.49&longitude=39.19"
       "&start_date=2025-06-01&end_date=2025-06-30"
       "&daily=temperature_2m_max"
       "&timezone=auto")

print(url)
````

The `?` starts the parameters and `&` separates them. That is all a request URL is.

Now send it. We wrap it in `try` / `except` so **a broken wifi connection does not break the lab**.

````python
try:
    with urllib.request.urlopen(url, timeout=20) as response:
        result = json.load(response)
    june = pd.DataFrame(result["daily"])
    print("SUCCESS - the API replied")
    display(june.head())
except Exception as e:
    june = None
    print(f"API unavailable ({type(e).__name__}) - carry on, this is not your fault")
````

**Thirty rows, fetched live, just now, from a server in Europe.**

> **Why this matters.** A file is frozen. An API is current: ask it tomorrow and you get tomorrow's
> data. That is how a deployed model gets the inputs it needs to make today's prediction.
>
> It also means an API can **fail** — the network drops, the service is busy, your quota runs out.
> Always handle that, as we did above.

**Real APIs add complications we are skipping today:** authentication keys, rate limits, and
pagination for large results. They exist; you will meet them; they are not today's lesson.

### Being a good citizen

- **Read the terms.** Open-Meteo is free for non-commercial use and asks for attribution.
- **Do not hammer it.** Ask once for a range of dates; do not loop ten thousand times.
- **Cache what you get.** Save the reply to a file rather than re-requesting the same thing.

That is exactly why our CSV exists: somebody asked the API once and saved the answer.

# Part 8 — Basic Statistics
*≈ 15 minutes*

You know the shape of the data. **Now describe it.**

We will compute the statistics **one at a time** first, so you know what each one means, and only

### Step 8.1 — One statistic at a time

Each of these takes a single column and returns a single number.

````python
weather["temp_max_c"].min()
````

````python
weather["temp_max_c"].max()
````

````python
weather["temp_max_c"].mean()
````

````python
weather["temp_max_c"].median()
````

````python
weather["temp_max_c"].std()
````

### Step 8.2 — What each one answers

| Statistic | The question it answers |
|---|---|
| **min** | What was the lowest value? |
| **max** | What was the highest value? |
| **mean** | What is the typical value, as an average? |
| **median** | What is the middle value — half above, half below? |
| **std** | How spread out are the values around the mean? |

**Read our numbers as a person, not a computer.** In Jeddah across 2023–2024:

- the coolest daily high was about **23 °C**, the hottest about **43 °C**
- the average daily high was about **33 °C**
- half of all days were above about **33 °C**
- a standard deviation of about **3.6 °C** means most days sat within a few degrees of that average

**Those are believable numbers for Jeddah.** That check is the whole point — you are looking for
anything *impossible*. A humidity of 150 %, a negative temperature, a maximum of 999.

### Step 8.3 — Mean or median?

Here they are close — about **32.8** and **33.3** — which tells you the distribution is roughly
symmetric.

> **When they differ a lot, believe the median.** The mean is dragged by extreme values; the
> median is not. If one day had been recorded as 999 °C, the mean would jump and the median would
> barely move.
>
> **Comparing the two is a free test for outliers.**

### Step 8.4 — All of them at once

Now that you know what each statistic means, here is the shortcut.

````python
weather.describe()
````

**Every statistic from Step 8.1, for every numeric column, in one command.**

Three rows are new:

| Row | Meaning |
|---|---|
| **count** | How many values were used — **missing ones are skipped**, so this can differ per column |
| **25 %** | A quarter of the values fall below this |
| **75 %** | Three quarters fall below this |

The 25 %, 50 % and 75 % rows are the **quartiles** — they mark the boundaries of the distribution.
The `50 %` row *is* the median.

> **`describe()` only shows numeric columns.** `date` is missing from that table because pandas

**The mean of a city name is meaningless.** For categorical columns you count instead.

| Column type | Statistics that make sense |
|---|---|
| **Numerical** | `mean()`, `median()`, `std()`, `describe()` |
| **Categorical** | `unique()`, `nunique()`, `value_counts()` |

Use the `sector` column from the demand table.

````python
demand["sector"].unique()
````

````python
demand["sector"].value_counts()
````

**One category, 731 rows.** Every row says `residential`, so this column is constant — exactly the

````python
stations["name"].value_counts()
````

Five names, one each — so `name` is an **identifier**, not a useful category. That is the

> **This is a foundational data science habit.** Before computing anything, ask: *is this column a

### Step 9.1 — `groupby` — the same statistic, per group

Do not think of `groupby` as syntax. Think of it as **a question**.

> **Question:** *What was the average maximum temperature in each year?*

To answer it we need a year column. Our `date` is text, so we convert it — the one small conversion

````python
weather["date"] = pd.to_datetime(weather["date"])

weather["year"] = weather["date"].dt.year

weather[["date", "year", "temp_max_c"]].head(3)
````

Now the question, in code. **Read it left to right as a sentence:**

```
weather                  take the weather table
.groupby("year")         split it into groups, one per year
["temp_max_c"]           from each group take the temperature column
.mean()                  and compute its average
```

````python
weather.groupby("year")["temp_max_c"].mean()
````

**2024 came out very slightly warmer than 2023.** Two years is far too short to call that a trend —
but notice that the *question* is now one line of code away.

> data?"* and then write the code. Students who learn `groupby` as syntax can reproduce the
> example; students who learn it as a question can answer their own.

Another question, same pattern:

> **Question:** *How many days did we record in each year?*

````python
weather.groupby("year")["temp_max_c"].count()
````

**365 and 366.** The data really does cover both years completely, and 2024 was a leap year.

### Step 9.2 — Correlation — a first look at relationships

Everything so far described **one column at a time**. Correlation is the first tool that looks at
**two columns together**.

It measures how strongly two columns move together, on a scale from **−1 to +1**:

$$ r \in [-1, 1] $$

| Value | Meaning |
|---|---|
| **+1** | Perfect: when one goes up, the other always goes up |
| **0** | No *linear* relationship |
| **−1** | Perfect opposite: one goes up, the other goes down |

````python
weather.corr(numeric_only=True).round(2)
````

**How to read this table.** Find a row, follow it across to a column, and that cell is their
correlation. The diagonal is all `1.00` because every column correlates perfectly with itself, and
the table is symmetric.

Three findings worth naming:

- **`temp_max_c` and `temp_min_c` ≈ 0.88** — very strong. A hot day has both a hot afternoon and a
  warm night. They measure related things.
- **`temp_max_c` and `solar_mj_m2` ≈ 0.58** — moderately strong. More sunshine, hotter day.
  Physically sensible.
- **`humidity_pct` and `temp_max_c` ≈ −0.12** — very weak and slightly negative. Essentially no
  linear relationship.

> ### Correlation is not causation.
>
> It says two columns move together. It does **not** say one causes the other — both may be driven
> by something else entirely, or the association may be coincidence.
>
> **And note the word *linear*.** Correlation near zero means no *straight-line* relationship. Two
> columns can be strongly related in a curved way and still score close to zero. That limitation is
> exactly why we plot.

> **`numeric_only=True`** tells pandas to skip the text columns rather than complain about them.

### One more thing in that table

**Look at the `year` row.** Every value is about `0.00`.

We created `year` a moment ago to answer a grouping question, and `corr()` treated it as a
quantity — because it is stored as `int64`, and pandas cannot see that it is really a label.

**This is exactly the trap from Part 5**, appearing in real output: *the meaning of a variable and
its pandas dtype are not the same thing*. The correlation between "temperature" and "which year it
was" is not a meaningful quantity, and you should ignore that row.

> **Nobody warned us.** pandas computed it, printed it, and formatted it neatly alongside the real
> correlations. **Only you know which columns are quantities.**

### Step 9.3 — One plot

Statistics summarise. **A plot shows you what the summary hides.**

The mean daily high was 32.8 °C — but a mean of 32.8 could come from many different shapes.
Only a picture says which one you have.

````python
import matplotlib.pyplot as plt

weather["temp_max_c"].hist(bins=30, figsize=(7, 3.5), color="#4D7EA8")

plt.title("Daily maximum temperature in Jeddah, 2023-2024")
plt.xlabel("maximum temperature (C)")
plt.ylabel("number of days")
plt.show()
````

**This is what `describe()` could not tell you.** The shape is not a single neat hump around 33 °C
— it is wide, slightly lopsided, with a long stretch of hot days above 35 °C and a thinner tail of
cooler ones.

> **Statistics and plots answer different questions.** `describe()` tells you *what the values
> are*. A plot tells you *how they are arranged*. **You need both, every time.**

We stop at one figure today. **Visualisation and full exploratory data analysis are Lab 3**, where
they get the time they deserve.

## Command map — your reference

Keep this. It is most of what you need for the next few weeks.

### Getting data in

| I want to… | Command |
|---|---|
| Load a CSV | `pd.read_csv("file.csv")` |
| Load Excel | `pd.read_excel("file.xlsx", sheet_name="Sheet1")` |
| Load JSON | `json.load(f)` then `pd.json_normalize(...)` |
| Load from a database | `pd.read_sql(query, connection)` |
| Load tables from a web page | `pd.read_html(url)` |
| Build a table by hand | `pd.DataFrame({"col": [1, 2, 3]})` |

### First look

| I want to… | Command |
|---|---|
| See the first rows | `df.head()` |
| Know rows × columns | `df.shape` |
| See column names | `df.columns` |
| Check data types | `df.dtypes` |
| Get a full summary | `df.info()` |
| Count distinct values | `df.nunique()` |
| Find missing values | `df.isna().sum()` |

### Statistics

| I want to… | Command |
|---|---|
| Select one column (**Series**) | `df["column"]` |
| Select several (**DataFrame**) | `df[["col1", "col2"]]` |
| Mean, median, std | `df["col"].mean()` · `.median()` · `.std()` |
| Min and max | `df["col"].min()` · `.max()` |
| All numeric statistics | `df.describe()` |
| List the categories | `df["col"].unique()` |
| Count the categories | `df["col"].value_counts()` |
| A statistic per group | `df.groupby("col")["other"].mean()` |
| Correlations | `df.corr(numeric_only=True)` |
| Convert to a NumPy array | `df[["a", "b"]].to_numpy()` |
